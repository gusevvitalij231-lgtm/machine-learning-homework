```python
import warnings
import time
from sklearn.exceptions import ConvergenceWarning

# Игнорируем только ConvergenceWarning
warnings.filterwarnings('ignore', category=ConvergenceWarning)
```

**Задание 2**

Используя функцию GridSearchCV подобрать гиперпараметры для Ridge, Lasso или ElasticNet. Сравнить качество улучшенной модели с LinearRegression.

```python
fold_n = 3

kf = KFold(n_splits = fold_n, shuffle = True, random_state = 14)

mse, mape, time_, r2 = 0, 0, 0, 0
for l1_ratio_ in [0.3, 0.5, 0.7]:
    for alpha_ in [0.8, 0.9, 1.0, 1.1, 1.2]:
        for train_index, test_index in kf.split(train_X, train_y):
            start = time.perf_counter()
            model = ElasticNet(random_state=14, alpha=alpha_, l1_ratio=l1_ratio_)
            model.fit(train_X.iloc[train_index], train_y.iloc[train_index])
            end = time.perf_counter()

            mse_ = mean_squared_error(train_y.iloc[test_index], model.predict(train_X.iloc[test_index]))
            mape_ = mean_absolute_percentage_error(train_y.iloc[test_index], model.predict(train_X.iloc[test_index]))
            r2_ = r2_score(train_y.iloc[test_index], model.predict(train_X.iloc[test_index]))
            mse += mse_
            mape += mape_
            r2 += r2_
            time_ += end - start

        print(f'Для alpha = {alpha_} и l1_ratio = {l1_ratio_}, MSE_mean = {mse/fold_n}, MAPE_mean = {mape/fold_n}, time_mean = {time_/fold_n:.4f}, r2_mean = {r2/fold_n}')

        mse, mape, r2, time_ = 0, 0, 0, 0
```

```python
model = ElasticNet(random_state=14)
param_grid = {'l1_ratio':[0.3, 0.4, 0.5, 0.7, 0.8], 'alpha':[0.8, 0.9, 1.0, 1.1, 1.2], 'fit_intercept':[True, False]}

grid_search = GridSearchCV(
    estimator=model,          
    param_grid=param_grid,     
    cv=5,                    
    scoring='neg_mean_squared_error',  
    verbose=1,                 
    n_jobs=-1                  
)

grid_search.fit(train_X, train_y)
```
